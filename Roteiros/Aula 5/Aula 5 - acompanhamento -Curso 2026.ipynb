## DataFrame e Series
os dataframes do pandas são como tabelas do excel e series são séries, sequencias unidimencional de algo, geralmente representa uma coluna do dataframe.
 

````python
#criando o primeiro dataframe
df = pd.DataFrame({'ar': [1,2,3], 'terra':[4,5,6]})
sr = pd.Series()

print(df, sr)
````

````python

#visualização de dados
df.head()
df.tail()
df.describe()
df.info()
df.shape
df.dtypes
````

````python

#seleção de dados
coluna = 'importante'
coluna1 = 'lembre de fazer'
coluna2 = 'depois'

df[coluna]
df[[coluna, coluna2]]
df.iloc[0]
df.loc[0, coluna]
df.at[0, coluna]
df.iat[0, 0]
````

````python

#filtragem de dados
valor ='depois'
valor2 = 'faz'
````

````python

df[df[coluna] > valor]
df[(df[coluna1] > valor1) & (df[coluna2] < valor2)]
````

````python

#plotagem de dados
df.plot.bar()
df.plot.line()
df.plot.scatter(x=coluna1, y=coluna2)
df.plot.hist()
````
